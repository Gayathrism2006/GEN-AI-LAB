```python
# Install libraries
!pip -q install transformers gtts pillow requests

from transformers import pipeline
from gtts import gTTS
from PIL import Image
from IPython.display import display, Audio
import requests

topic = "The benefits of renewable energy"

# Text Generation
generator = pipeline("text2text-generation", model="google/flan-t5-base")

text = generator(
    f"Write a short paragraph about {topic}",
    max_length=80
)[0]["generated_text"]

print(text)

# Download a free placeholder image
url = "https://picsum.photos/512"

img = Image.open(requests.get(url, stream=True).raw)

img.save("content_image.png")

display(img)

# Audio Generation
tts = gTTS(text=text, lang="en")
tts.save("content_audio.mp3")

display(Audio("content_audio.mp3"))
```
