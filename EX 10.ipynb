```python
# ==========================================================
# DistilBERT Sentiment Analysis using Hugging Face Pipeline
# Google Colab Compatible
# ==========================================================

# Install required library
!pip install -q transformers torch

# Import pipeline
from transformers import pipeline

# Load pre-trained sentiment analysis model
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Sample reviews
reviews = [
    "This movie was fantastic! I loved every minute of it.",
    "The acting was terrible and the story was boring.",
    "It was an average movie with some good scenes.",
    "Absolutely amazing! Highly recommended.",
    "Worst movie I have ever watched."
]

# Predict sentiment
for review in reviews:
    result = classifier(review)
    print("Review :", review)
    print("Prediction :", result[0]["label"])
    print("Confidence :", round(result[0]["score"], 4))
    print("-" * 60)
```
